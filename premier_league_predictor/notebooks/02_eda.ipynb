# Exploratory Data Analysis (EDA)

En este notebook exploraremos los datos crudos que descargamos de la historia de los partidos de la Premier League.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")

import warnings
warnings.filterwarnings('ignore')

# Load data
df = pd.read_csv('../data/raw/premier_league_raw.csv')
df.head()
```

## 1. Análisis de la Variable Objetivo (`FTR`)

Queremos ver el balance de victorias del Local (H), Empate (D), y Visita (A).

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='FTR', order=['H','D','A'], palette="viridis")
plt.title('Distribución de Resultados Finales (FTR)')
plt.show()

# Porcentajes
df['FTR'].value_counts(normalize=True) * 100
```

Usualmente en el fútbol existe un desbalance natural a favor del Local (Home Advantage).
