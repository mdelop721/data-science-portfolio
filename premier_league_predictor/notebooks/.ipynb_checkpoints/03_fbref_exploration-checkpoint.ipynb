# Exploración de Datos Avanzados con FBref (Opta)

Vamos a utilizar la librería `soccerdata` que acabamos de instalar para conectarnos al portal FBref y observar la densidad de las métricas tácticas avanzadas que provee Opta para la temporada 2023-2024 de la Premier League.

```python
import pandas as pd
import soccerdata as sd

# Set options para visualizar todas las columnas en pandas
pd.set_option('display.max_columns', None)

import warnings
warnings.filterwarnings('ignore')
```

### 1. Inicializar el Scraper (La Clase FBref)
Le decimos explícitamente a la clase qué liga y qué temporada vamos a mapear. Esto descarga el caché en segundo plano de manera limpia sin re-consultar los servidores de la página.

```python
# ENG-Premier League, temporada de agosto 2023 a mayo 2024 ('2324')
fbref = sd.FBref(leagues="ENG-Premier League", seasons="2324")
print("Scraper de FBref inicializado con éxito")
```

### 2. Exploración del Calendario General (The Schedule)
El primer DataFrame nos da información sobre todos los partidos de la liga, incluyendo los valiosísimos xG (Goles Esperados) de cada encuentro.

```python
df_schedule = fbref.read_schedule()
df_schedule.head()
```

### 3. Estadísticas de Equipo Estructurales (Team Match Stats)
FBref cataloga el rendimiento en varios DataFrames gigantes. Intentemos extraer las **estadísticas de tiro** (`shooting`) por partido, que revelan la verdadera presión defensiva/ofensiva sin importar si fue gol:

```python
# Tipos de estadísitcas disponibles en FBref: 
# 'summary', 'keepers', 'passing', 'passing_types', 'defense', 'possession', 'misc', etc.

df_shooting = fbref.read_team_match_stats(stat_type="shooting")
df_shooting.head()
```

### 4. Estadísticas de Posesión y Control Activo (`possession`)
¿Qué equipo dominó la pelota en el último tercio de la cancha?

```python
df_possession = fbref.read_team_match_stats(stat_type="possession")
df_possession.head()
```
