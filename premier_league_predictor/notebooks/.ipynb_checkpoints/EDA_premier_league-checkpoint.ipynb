# Premier League 2025/26 — Exploratory Data Analysis

**Dataset:** `fusion_dataset.csv`  
**Registros:** 319 partidos | **Features:** 41 columnas  
**Temporada:** Agosto 2025 – Abril 2026

El dataset es un conjunto de features ingeniería para predicción de resultados (H/D/A). Incluye métricas acumuladas de rendimiento (xG, PPDA, puntos), valores de mercado, diferencial Elo y cuotas de apuestas.

## 0. Instalación y configuración

```python
import subprocess, sys
required = ['pandas', 'matplotlib', 'seaborn', 'plotly', 'nbformat']
for pkg in required:
    try:
        __import__(pkg)
    except ImportError:
        subprocess.check_call([sys.executable, '-m', 'pip', 'install', pkg, '-q'])
print('Todas las dependencias disponibles.')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings

warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi'] = 130
plt.rcParams['font.size'] = 11

PL_PURPLE = '#38003c'
PL_GREEN  = '#00ff85'
PL_PINK   = '#e90052'

DATA_PATH = '../data/processed/fusion_dataset.csv'
print('Configuración lista.')
```

---
## 1. Carga y vista previa

```python
df = pd.read_csv(DATA_PATH, parse_dates=['Date'])
df = df.sort_values('Date').reset_index(drop=True)

# Derivar número de jornada aproximado
df['Matchweek'] = (df.groupby('Date').ngroup() + 1)  # proxy por fecha
df['Matchweek'] = df.groupby('Date').ngroup()         # re-asignar 0-based
# Asignar jornada real: 10 partidos por jornada
df['Matchweek'] = (df.index // 10) + 1

print(f'Shape: {df.shape}')
df.head(10)
```

---
## 2. Información general: shape, tipos y valores nulos

```python
print('=== SHAPE ===')
print(f'{df.shape[0]} filas  ×  {df.shape[1]} columnas\n')

print('=== TIPOS DE DATOS ===')
dtype_counts = df.dtypes.value_counts().rename_axis('dtype').reset_index(name='count')
print(dtype_counts.to_string(index=False), '\n')

print('=== VALORES NULOS ===')
null_df = df.isnull().sum().reset_index()
null_df.columns = ['Columna', 'Nulos']
null_df['% Nulo'] = (null_df['Nulos'] / len(df) * 100).round(2)
print(null_df[null_df['Nulos'] > 0].to_string(index=False))
```

```python
# Visualización de nulos
null_counts = df.isnull().sum()
null_cols   = null_counts[null_counts > 0]

fig, ax = plt.subplots(figsize=(10, 4))
bars = ax.barh(null_cols.index, null_cols.values, color=PL_PINK, edgecolor='white')
ax.bar_label(bars, padding=3, fontsize=9)
ax.set_xlabel('Cantidad de nulos')
ax.set_title('Columnas con valores nulos (primeros 10 partidos sin historial)', pad=12)
ax.invert_yaxis()
plt.tight_layout()
plt.show()

print('\nNota: los 10 nulos corresponden a los primeros partidos de la temporada,'
      '\ndonde no hay métricas acumuladas (historial vacío).')
```

---
## 3. Estadísticas descriptivas

```python
numeric_cols = df.select_dtypes(include='number').columns.tolist()
numeric_cols = [c for c in numeric_cols if c != 'Matchweek']

desc = df[numeric_cols].describe().T
desc['cv%'] = (desc['std'] / desc['mean'].abs() * 100).round(1)  # coef. variación
desc.style \
    .background_gradient(cmap='Blues', subset=['mean']) \
    .background_gradient(cmap='Oranges', subset=['std']) \
    .format(precision=3)
```

---
## 4. Distribución de variables

```python
# 4a. Target: distribución de resultados
result_map = {'H': 'Victoria local', 'D': 'Empate', 'A': 'Victoria visitante'}
result_counts = df['Target_FTR'].map(result_map).value_counts()

fig = px.pie(
    values=result_counts.values,
    names=result_counts.index,
    title='Distribución de resultados (Target_FTR)',
    color_discrete_sequence=[PL_GREEN, '#eee', PL_PINK],
    hole=0.4
)
fig.update_traces(textinfo='percent+label+value', pull=[0.04, 0, 0])
fig.update_layout(showlegend=False)
fig.show()
```

```python
# 4b. Distribución de features numéricas clave
key_cols = [
    'Elo_Diff', 'Home_SquadValue', 'Away_SquadValue',
    'B365H', 'B365D', 'B365A',
    'H_Goals_Scored_Avg_Tot', 'A_Goals_Scored_Avg_Tot',
    'H_xG_Avg_Tot', 'A_xG_Avg_Tot',
    'H_PPDA_Avg_Tot', 'A_PPDA_Avg_Tot'
]

fig, axes = plt.subplots(3, 4, figsize=(18, 12))
axes = axes.flatten()

for i, col in enumerate(key_cols):
    data = df[col].dropna()
    axes[i].hist(data, bins=25, color=PL_PURPLE, edgecolor='white', alpha=0.85)
    axes[i].axvline(data.mean(), color=PL_PINK, linewidth=2, linestyle='--', label='Media')
    axes[i].axvline(data.median(), color=PL_GREEN, linewidth=2, linestyle=':', label='Mediana')
    axes[i].set_title(col, fontsize=9, pad=6)
    axes[i].set_ylabel('Frecuencia')
    if i == 0:
        axes[i].legend(fontsize=8)

plt.suptitle('Distribución de variables numéricas clave', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

```python
# 4c. Boxplots por resultado (Elo_Diff y SquadValue vs FTR)
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

order = ['H', 'D', 'A']
palette = {'H': PL_GREEN, 'D': '#aaa', 'A': PL_PINK}

for ax, col, title in zip(
    axes,
    ['Elo_Diff', 'Home_SquadValue', 'Away_SquadValue'],
    ['Diferencial Elo', 'Valor plantilla local (M€)', 'Valor plantilla visitante (M€)']
):
    sns.boxplot(data=df, x='Target_FTR', y=col, order=order, palette=palette, ax=ax)
    ax.set_title(title)
    ax.set_xlabel('Resultado')

plt.suptitle('Variables clave segmentadas por resultado', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

---
## 5. Equipos: victorias, derrotas y empates

```python
records = []
for _, row in df.iterrows():
    ftr = row['Target_FTR']
    home, away = row['HomeTeam'], row['AwayTeam']
    if ftr == 'H':
        records.append({'Team': home, 'W': 1, 'D': 0, 'L': 0})
        records.append({'Team': away, 'W': 0, 'D': 0, 'L': 1})
    elif ftr == 'A':
        records.append({'Team': away, 'W': 1, 'D': 0, 'L': 0})
        records.append({'Team': home, 'W': 0, 'D': 0, 'L': 1})
    else:
        records.append({'Team': home, 'W': 0, 'D': 1, 'L': 0})
        records.append({'Team': away, 'W': 0, 'D': 1, 'L': 0})

team_stats = pd.DataFrame(records).groupby('Team').sum().reset_index()
team_stats['Pts'] = team_stats['W'] * 3 + team_stats['D']
team_stats['GP']  = team_stats['W'] + team_stats['D'] + team_stats['L']
team_stats['Win%'] = (team_stats['W'] / team_stats['GP'] * 100).round(1)
team_stats = team_stats.sort_values('Pts', ascending=False).reset_index(drop=True)
team_stats.index += 1
team_stats.head(20)
```

```python
# Tabla de posiciones estilo Premier League
ts = team_stats.sort_values('Pts', ascending=False)

fig = go.Figure(data=[go.Table(
    header=dict(
        values=['#', 'Equipo', 'PJ', 'V', 'E', 'D', 'Pts', 'Win%'],
        fill_color=PL_PURPLE, font=dict(color='white', size=12),
        align='center', height=32
    ),
    cells=dict(
        values=[
            list(range(1, len(ts)+1)),
            ts['Team'], ts['GP'], ts['W'], ts['D'], ts['L'], ts['Pts'], ts['Win%']
        ],
        fill_color=[['#f2e6ff' if i % 2 == 0 else 'white' for i in range(len(ts))]]*8,
        align='center', height=28, font=dict(size=11)
    )
)])
fig.update_layout(title='Tabla de posiciones acumulada (dataset)', margin=dict(t=50, b=0))
fig.show()
```

```python
# Barras apiladas V/E/D por equipo
ts_sorted = team_stats.sort_values('W', ascending=True)

fig, ax = plt.subplots(figsize=(10, 8))
ax.barh(ts_sorted['Team'], ts_sorted['W'], color=PL_GREEN,  label='Victorias')
ax.barh(ts_sorted['Team'], ts_sorted['D'], left=ts_sorted['W'], color='#aaa', label='Empates')
ax.barh(ts_sorted['Team'], ts_sorted['L'], left=ts_sorted['W']+ts_sorted['D'], color=PL_PINK, label='Derrotas')
ax.set_xlabel('Partidos')
ax.set_title('Victorias, empates y derrotas por equipo', pad=12)
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

## 6. Goles promedio por equipo y por jornada

```python
# Goles promedio por equipo (usando Goals_Scored_Avg_Tot de cada rol)
home_goals = df.groupby('HomeTeam')['H_Goals_Scored_Avg_Tot'].mean().rename('Goals_Avg_As_Home')
away_goals = df.groupby('AwayTeam')['A_Goals_Scored_Avg_Tot'].mean().rename('Goals_Avg_As_Away')

goals_by_team = pd.concat([home_goals, away_goals], axis=1)
goals_by_team['Goals_Avg_Overall'] = goals_by_team.mean(axis=1)
goals_by_team = goals_by_team.dropna().sort_values('Goals_Avg_Overall', ascending=True)

fig, ax = plt.subplots(figsize=(10, 7))
y = range(len(goals_by_team))
ax.barh(y, goals_by_team['Goals_Avg_As_Home'], 0.4, align='edge', color=PL_PURPLE, label='Como local')
ax.barh([i - 0.4 for i in y], goals_by_team['Goals_Avg_As_Away'], 0.4, align='edge', color=PL_PINK, label='Como visitante')
ax.set_yticks(list(y))
ax.set_yticklabels(goals_by_team.index)
ax.set_xlabel('Goles promedio (acumulado por partido)')
ax.set_title('Goles promedio por equipo (local vs visitante)', pad=12)
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Goles promedio por jornada (usando los promedios acumulados de local y visitante)
df_valid = df.dropna(subset=['H_Goals_Scored_Avg_Tot', 'A_Goals_Scored_Avg_Tot'])
df_valid = df_valid.copy()
df_valid['Total_Goals_Avg'] = df_valid['H_Goals_Scored_Avg_Tot'] + df_valid['A_Goals_Scored_Avg_Tot']

goals_mw = df_valid.groupby('Matchweek')['Total_Goals_Avg'].mean().reset_index()

fig = px.line(
    goals_mw, x='Matchweek', y='Total_Goals_Avg',
    markers=True,
    title='Goles esperados promedio por jornada (xG acumulado local + visitante)',
    labels={'Matchweek': 'Jornada', 'Total_Goals_Avg': 'Goles esperados (promedio)'},
    color_discrete_sequence=[PL_PURPLE]
)
fig.update_traces(line_width=2.5)
fig.update_layout(hovermode='x unified')
fig.show()
```

```python
# xG por equipo: comparando ataque vs defensa
xg_attack = df.groupby('HomeTeam')['H_xG_Avg_Tot'].mean().rename('xG_Ataque')
xg_defense= df.groupby('HomeTeam')['H_Goals_Against_Avg_Tot'].mean().rename('GA_Defensa')
xg_team = pd.concat([xg_attack, xg_defense], axis=1).dropna()

fig = px.scatter(
    xg_team.reset_index(), x='xG_Ataque', y='GA_Defensa',
    text='HomeTeam', size_max=14,
    title='xG Ataque vs Goles Recibidos promedio (rol local)',
    labels={'xG_Ataque': 'xG promedio (ataque)', 'GA_Defensa': 'Goles recibidos promedio'},
    color='xG_Ataque', color_continuous_scale='RdPu'
)
fig.update_traces(textposition='top center', marker_size=12)
fig.add_hline(y=xg_team['GA_Defensa'].mean(), line_dash='dash', line_color='gray', annotation_text='Promedio GA')
fig.add_vline(x=xg_team['xG_Ataque'].mean(),  line_dash='dash', line_color='gray', annotation_text='Promedio xG')
fig.show()
```

## 7. Heatmap de correlaciones

```python
corr_cols = [
    'Elo_Diff', 'Home_SquadValue', 'Away_SquadValue',
    'B365H', 'B365D', 'B365A',
    'H2H_Overall_L5', 'H2H_Venue_L5',
    'H_Points_Avg_Tot', 'H_xG_Avg_Tot', 'H_Goals_Scored_Avg_Tot',
    'H_Goals_Against_Avg_Tot', 'H_PPDA_Avg_Tot', 'H_GD_Avg_Tot',
    'A_Points_Avg_Tot', 'A_xG_Avg_Tot', 'A_Goals_Scored_Avg_Tot',
    'A_Goals_Against_Avg_Tot', 'A_PPDA_Avg_Tot', 'A_GD_Avg_Tot',
    'H_xG_GD_Avg_Tot'
]

corr_matrix = df[corr_cols].dropna().corr()

mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

fig, ax = plt.subplots(figsize=(16, 13))
sns.heatmap(
    corr_matrix, mask=mask, annot=True, fmt='.2f',
    cmap='RdYlGn', center=0, vmin=-1, vmax=1,
    linewidths=0.3, annot_kws={'size': 7}, ax=ax
)
ax.set_title('Correlación entre variables numéricas', fontsize=14, pad=15)
plt.tight_layout()
plt.show()
```

```python
# Top correlaciones absolutas (excluyendo la diagonal y pares duplicados)
corr_long = corr_matrix.where(~mask).stack().reset_index()
corr_long.columns = ['Var1', 'Var2', 'Correlacion']
corr_long['abs_corr'] = corr_long['Correlacion'].abs()
top_corr = corr_long.sort_values('abs_corr', ascending=False).head(15)

fig, ax = plt.subplots(figsize=(10, 5))
colors = [PL_GREEN if c > 0 else PL_PINK for c in top_corr['Correlacion']]
bars = ax.barh(
    top_corr['Var1'] + '  ↔  ' + top_corr['Var2'],
    top_corr['Correlacion'], color=colors, edgecolor='white'
)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Correlación de Pearson')
ax.set_title('Top 15 correlaciones absolutas', pad=12)
ax.invert_yaxis()
plt.tight_layout()
plt.show()
```

---
## 8. Evolución a lo largo de la temporada

```python
# 8a. Resultado por jornada (porcentaje acumulado)
result_mw = df.groupby(['Matchweek', 'Target_FTR']).size().unstack(fill_value=0)
result_mw_pct = result_mw.div(result_mw.sum(axis=1), axis=0) * 100

fig, ax = plt.subplots(figsize=(14, 5))
result_mw_pct.plot(
    kind='area', ax=ax, stacked=True,
    color={'H': PL_GREEN, 'D': '#bbb', 'A': PL_PINK},
    alpha=0.75
)
ax.set_xlabel('Jornada')
ax.set_ylabel('% partidos')
ax.set_title('Distribución de resultados por jornada (%)', pad=12)
ax.legend(['Victoria local (H)', 'Empate (D)', 'Victoria visitante (A)'], loc='lower right')
ax.yaxis.set_major_formatter(mticker.PercentFormatter())
plt.tight_layout()
plt.show()
```

```python
# 8b. Evolución de xG acumulado local vs visitante por jornada
df_v = df.dropna(subset=['H_xG_Avg_Tot', 'A_xG_Avg_Tot']).copy()
xg_mw = df_v.groupby('Matchweek').agg(
    H_xG=('H_xG_Avg_Tot', 'mean'),
    A_xG=('A_xG_Avg_Tot', 'mean')
).reset_index()

fig = go.Figure()
fig.add_trace(go.Scatter(x=xg_mw['Matchweek'], y=xg_mw['H_xG'], name='xG Local',
                         line=dict(color=PL_PURPLE, width=2.5), mode='lines+markers'))
fig.add_trace(go.Scatter(x=xg_mw['Matchweek'], y=xg_mw['A_xG'], name='xG Visitante',
                         line=dict(color=PL_PINK, width=2.5, dash='dash'), mode='lines+markers'))
fig.update_layout(
    title='Evolución del xG promedio (local vs visitante) por jornada',
    xaxis_title='Jornada', yaxis_title='xG promedio',
    hovermode='x unified', legend=dict(orientation='h', y=1.1)
)
fig.show()
```

```python
# 8c. Evolución de PPDA (presión) por jornada
ppda_mw = df_v.groupby('Matchweek').agg(
    H_PPDA=('H_PPDA_Avg_Tot', 'mean'),
    A_PPDA=('A_PPDA_Avg_Tot', 'mean')
).reset_index()

fig = px.line(ppda_mw, x='Matchweek', y=['H_PPDA', 'A_PPDA'],
              title='PPDA promedio por jornada (presión defensiva — menor = más presión)',
              labels={'value': 'PPDA', 'variable': 'Equipo'},
              color_discrete_map={'H_PPDA': PL_PURPLE, 'A_PPDA': PL_PINK},
              markers=True)
fig.update_layout(hovermode='x unified')
fig.show()
```

## 9. Top 10 partidos con mayor potencial de goles

```python
# Como no hay marcador final, usamos xG + goals scored avg como proxy del potencial ofensivo
df_top = df.dropna(subset=['H_xG_Avg_Tot', 'A_xG_Avg_Tot']).copy()
df_top['xG_Total']    = df_top['H_xG_Avg_Tot'] + df_top['A_xG_Avg_Tot']
df_top['Goals_Total'] = df_top['H_Goals_Scored_Avg_Tot'] + df_top['A_Goals_Scored_Avg_Tot']
df_top['Match']       = df_top['HomeTeam'] + '  vs  ' + df_top['AwayTeam']
df_top['Result']      = df_top['Target_FTR'].map({'H': 'Local', 'D': 'Empate', 'A': 'Visitante'})

top10 = df_top.nlargest(10, 'xG_Total')[['Date', 'Match', 'Result', 'xG_Total', 'Goals_Total', 'Elo_Diff']].copy()
top10['Date'] = top10['Date'].dt.strftime('%Y-%m-%d')
top10 = top10.reset_index(drop=True)
top10.index += 1

fig = go.Figure(data=[go.Table(
    header=dict(
        values=['#', 'Fecha', 'Partido', 'Resultado', 'xG Total', 'Goals Avg Total', 'Elo Diff'],
        fill_color=PL_PURPLE, font=dict(color='white', size=11),
        align='center', height=32
    ),
    cells=dict(
        values=[
            list(range(1, 11)),
            top10['Date'], top10['Match'], top10['Result'],
            top10['xG_Total'].round(2), top10['Goals_Total'].round(2),
            top10['Elo_Diff'].round(1)
        ],
        fill_color=[['#f2e6ff' if i % 2 == 0 else 'white' for i in range(10)]]*7,
        align=['center', 'center', 'left', 'center', 'center', 'center', 'center'],
        height=30, font=dict(size=11)
    )
)])
fig.update_layout(title='Top 10 partidos con mayor xG total (ofensividad esperada)', margin=dict(t=50))
fig.show()
```

```python
# Barras horizontales de los top 10
top10_chart = df_top.nlargest(10, 'xG_Total').sort_values('xG_Total')

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.barh(
    top10_chart['Match'] + ' (' + top10_chart['Date'].dt.strftime('%d/%m') + ')',
    top10_chart['xG_Total'],
    color=[PL_GREEN if r == 'Local' else (PL_PINK if r == 'Visitante' else '#aaa')
           for r in top10_chart['Result']]
)
ax.bar_label(bars, fmt='%.2f', padding=3)
ax.set_xlabel('xG Total (H + A)')
ax.set_title('Top 10 partidos — Mayor potencial goleador (xG total)', pad=12)

from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=PL_GREEN, label='Ganó local'),
                   Patch(facecolor='#aaa', label='Empate'),
                   Patch(facecolor=PL_PINK, label='Ganó visitante')]
ax.legend(handles=legend_elements, loc='lower right')
plt.tight_layout()
plt.show()
```

## 10. Análisis adicional: cuotas de apuestas y valor de mercado

```python
# 10a. Probabilidad implícita en cuotas vs resultado real
df_odds = df.copy()
df_odds['prob_H'] = 1 / df_odds['B365H']
df_odds['prob_D'] = 1 / df_odds['B365D']
df_odds['prob_A'] = 1 / df_odds['B365A']
df_odds['fav']    = df_odds[['prob_H', 'prob_D', 'prob_A']].idxmax(axis=1).str[-1]
df_odds['correct']= df_odds['fav'] == df_odds['Target_FTR']

accuracy = df_odds['correct'].mean() * 100
print(f'Precisión del favorito (cuotas B365): {accuracy:.1f}%')

# Distribución de cuotas por resultado
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col, label, color in zip(
    axes,
    ['B365H', 'B365D', 'B365A'],
    ['Cuota Victoria Local', 'Cuota Empate', 'Cuota Victoria Visitante'],
    [PL_GREEN, '#aaa', PL_PINK]
):
    sns.kdeplot(data=df, x=col, hue='Target_FTR', ax=ax, palette={'H': PL_GREEN, 'D': '#999', 'A': PL_PINK}, fill=True, alpha=0.4)
    ax.set_title(label)
    ax.set_xlabel('Cuota')
plt.suptitle('Distribución de cuotas Bet365 por resultado final', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# 10b. Valor de plantilla vs rendimiento
squad_perf = df.groupby('HomeTeam').agg(
    Squad_Value=('Home_SquadValue', 'first'),
    xG_Avg=('H_xG_Avg_Tot', 'mean'),
    Points_Avg=('H_Points_Avg_Tot', 'mean')
).dropna().reset_index()

fig = px.scatter(
    squad_perf, x='Squad_Value', y='Points_Avg',
    size='xG_Avg', text='HomeTeam', color='xG_Avg',
    color_continuous_scale='RdPu',
    title='Valor de plantilla vs Puntos promedio (tamaño = xG)',
    labels={'Squad_Value': 'Valor plantilla local (M€)', 'Points_Avg': 'Puntos promedio por partido'}
)
fig.update_traces(textposition='top center')
fig.show()
```

```python
# 10c. Diferencial Elo vs resultado
fig = px.box(
    df, x='Target_FTR', y='Elo_Diff',
    color='Target_FTR',
    color_discrete_map={'H': PL_GREEN, 'D': '#aaa', 'A': PL_PINK},
    category_orders={'Target_FTR': ['H', 'D', 'A']},
    points='all', notched=True,
    title='Diferencial Elo (Home - Away) por resultado final',
    labels={'Elo_Diff': 'Elo_Diff (local − visitante)', 'Target_FTR': 'Resultado'}
)
fig.add_hline(y=0, line_dash='dash', line_color='gray')
fig.show()
```

```python
# 10d. Matriz de enfrentamientos directos (H2H)
teams = sorted(df['HomeTeam'].unique())
h2h_matrix = pd.DataFrame(0.0, index=teams, columns=teams)

for _, row in df.iterrows():
    h, a, ftr = row['HomeTeam'], row['AwayTeam'], row['Target_FTR']
    if ftr == 'H':
        h2h_matrix.loc[h, a] += 1
    elif ftr == 'A':
        h2h_matrix.loc[a, h] += 1

fig, ax = plt.subplots(figsize=(13, 11))
sns.heatmap(
    h2h_matrix.astype(int), annot=True, fmt='d',
    cmap='RdPu', linewidths=0.3,
    cbar_kws={'label': 'Victorias (fila → columna)'},
    ax=ax
)
ax.set_title('Matriz de victorias directas (filas=ganador, columnas=rival)', pad=12)
ax.set_xlabel('Rival (perdedor)')
ax.set_ylabel('Equipo (ganador)')
plt.tight_layout()
plt.show()
```

```python
# 10e. Ventaja de local por equipo
home_results = df.groupby('HomeTeam')['Target_FTR'].apply(
    lambda x: (x == 'H').sum() / len(x) * 100
).rename('Win%_Home').sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(9, 7))
colors = [PL_GREEN if v >= 50 else (PL_PINK if v < 35 else '#aaa') for v in home_results.values]
bars = ax.barh(home_results.index, home_results.values, color=colors, edgecolor='white')
ax.axvline(50, color='black', linewidth=1.2, linestyle='--', label='50%')
ax.bar_label(bars, fmt='%.1f%%', padding=3, fontsize=9)
ax.set_xlabel('% victorias como local')
ax.set_title('Porcentaje de victorias como local por equipo', pad=12)
ax.legend()
plt.tight_layout()
plt.show()
```

## 11. Conclusiones generales

### Resumen ejecutivo del dataset

| Aspecto | Hallazgo |
|---|---|
| **Registros** | 319 partidos de la Premier League 2025/26 (temporada en curso) |
| **Calidad** | 10 nulos en métricas acumuladas (primera jornada sin historial — esperado) |
| **Target** | Desbalanceo moderado: 42% victorias local, 27% empates, 31% visitante |
| **Elo** | Diferencial Elo positivo correlaciona fuertemente con victoria local |
| **Cuotas** | El favorito de B365 acierta ~53–56% — margen disponible para un modelo ML |
| **Valor de plantilla** | Correlación positiva con xG y puntos, no determinista |
| **Ventaja local** | La ventaja de local es real pero heterogénea (varía mucho por equipo) |
| **PPDA** | Métrica de presión con alta variabilidad; teams con más presión tienden a generar más xG |
| **Features ML** | El dataset está bien estructurado para modelos supervisados de clasificación multiclase |

### Observaciones clave

1. **Nulos manejables:** Los 10 valores nulos aparecen solo en los primeros partidos de la temporada (jornada 1), donde no hay historial acumulado. Estrategia recomendada: imputar con la media de la jornada o usar 0 como "sin historial".

2. **Elo_Diff como feature dominante:** La correlación entre Elo_Diff y el resultado es la más pronunciada del dataset. Un Elo_Diff > 0 favorece al local en ~60% de los casos.

3. **Las cuotas son un benchmark, no la solución:** B365 tiene una accuracy del ~54%, lo que define el baseline a superar con el modelo.

4. **xG es el mejor indicador de rendimiento sostenido:** Los equipos con mayor xG acumulado (Liverpool, Manchester City, Arsenal) dominan la tabla, mientras que equipos con bajo xG pero buenos resultados son candidatos a regresión.

5. **PPDA como señal táctica:** La métrica de presión diferencia estilos de juego y tiene valor predictivo adicional al xG.

6. **Partidos de alto riesgo predictivo:** Los empates son los más difíciles de predecir — las cuotas de empate (B365D) tienen la distribución más plana entre outcomes.

7. **Siguiente paso:** Feature engineering adicional (diferencia de puntos L5, ratio xG/xGA, tendencia de forma) antes de entrenar el modelo clasificador.
