# Understat: The Home of xG (Goles Esperados)

Understat es una plataforma especializada que calcula métricas hiper-avanzadas (xG, xPTS, PPDA) de manera libre y gratuita. `soccerdata` la incluye como un módulo paralelo a `FBref`.
Veamos cómo inicializarlo y extraer sus secretos.

```python
import pandas as pd
import soccerdata as sd
import warnings
warnings.filterwarnings('ignore')

pd.set_option('display.max_columns', None)
```

### 1. Inicializar la clase `Understat`
Exactamente igual que con FBref, le decimos el país-liga y el código de la temporada (Agosto 2023 - Mayo 2024).

```python
understat = sd.Understat(leagues="ENG-Premier League", seasons="2324")
print("Scraper de Understat listo")
```

### 2. Recuperar el Match Log (Reporte de Partidos)
A diferencia de FBref que dividía todo en 20 tablas, ¡Understat nos entrega todo empacado en un solo DataFrame colosal sin MultiIndexes! (Una maravilla para el Data Science).

```python
df_understat = understat.read_team_match_stats()

# Imprimimos las columnas para que las conozcas
print("Columnas disponibles:")
print(list(df_understat.columns))

# Vemos las primeras 5 filas
df_understat.head()
```

### 3. Entendiendo el Lenguaje de Understat

Bajo la estructura de este dataframe vamos a encontrar columnas clave como:
* `home_xg` y `away_xg`: Los Goles Esperados.
* `home_np_xg`: Non-Penalty Expected Goals (El xG real de la ofensiva natural, sin la trampa de un tiro penal).
* `home_ppda`: Pases Permitidos por Acción Defensiva. *(Ej. Si el PPDA es 6, significa que presiono tanto que el rival solo puede dar 6 pases antes de que yo le quite el balón o haga falta)*.
* `home_deep_completions`: Cuántos pases completé muy adentro de la zona de mi rival.
